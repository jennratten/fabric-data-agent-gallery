## Fabric Data Agent | Ground Truth Dataset Generator

Reusable notebook template for creating **ground truth evaluation datasets** for Fabric data agents over any Power BI / Fabric semantic model, including Direct Lake semantic models.

It reviews semantic model metadata, selects visible explicit measures and useful dimensions, generates benchmark questions, executes validation DAX queries, and writes:

- `*sm__{SEMANTIC_MODEL_NAME}_ground_truth.csv`
- `*sm__{SEMANTIC_MODEL_NAME}_ground_truth.json`

##### Documentation: 
https://learn.microsoft.com/en-us/fabric/data-science/evaluate-data-agent

#### Recommendation

Before formal evaluation, review the generated questions and validation queries with the model owner or SME. Remove tests that rely on ambiguous metadata, duplicate measures, or fields that should not be exposed to an agent. For answer scoring, add columns such as `agent_answer`, `is_exact_match`, `is_numerically_equivalent`, `allowed_tolerance`, `failure_category`, `reviewer`, and `reviewed_utc`.

#### Step 1 – Configure which semantic model to profile

This notebook generates **ground truth test data** for a Fabric data agent by querying a Power BI / Fabric semantic model.

In this step you:

- Tell the notebook which semantic model (dataset) to use
- Optionally specify the workspace name
- Control how many tests and fields are included in the output

#### What you need to change

- `SEMANTIC_MODEL_NAME` – set this to the exact name of your semantic model as it appears in the Fabric workspace.
- `WORKSPACE_NAME` – usually you can leave this as `None` if the model is in the **same workspace** as this notebook. If the model is in another workspace you have access to, set this to that workspace name.

You can also adjust the limits if you want **more or fewer tests**:

- `MAX_TESTS` – maximum number of ground‑truth questions to generate
- `MAX_MEASURES_FOR_KPIS` – how many key measures to focus on
- `MAX_DIMENSION_COLUMNS` – how many dimension columns to profile
- `MAX_DISTINCT_VALUES_FOR_DIMENSION` – skip dimensions with more than this many distinct values (to avoid huge tables)
- `TOP_N` – how many items to include in “Top N / Bottom N” style questions

After you update these values, **run the next code cell** to apply your configuration.


````python
# ============================================================
# 1. Configuration
# ============================================================
SEMANTIC_MODEL_NAME = "AdventureWorks_OfficeSupplies"  # <-- change this to your semantic model name
WORKSPACE_NAME = None  # <-- add the workspace name; leave this set to None to return the current workspace
OUTPUT_BASENAME = f'sm__{SEMANTIC_MODEL_NAME}_ground_truth' 
OUTPUT_FOLDER = "Files/ground_truth" # "."
MAX_TESTS = 50 
MAX_MEASURES_FOR_KPIS = 5
MAX_DIMENSION_COLUMNS = 12
MAX_DISTINCT_VALUES_FOR_DIMENSION = 50
TOP_N = 3
print("Semantic model:", SEMANTIC_MODEL_NAME)
````

## Step 2 – Import libraries and define helper functions

This step loads Python libraries and defines helper functions used throughout the notebook.

You **do not usually need to change anything in this cell**.

### What this cell does

- Imports general‑purpose Python modules like `json`, `os`, `re`, and `datetime`
- Imports **pandas** for working with tabular data
- Imports `sempy.fabric` so we can call Fabric / Power BI semantic model APIs (for example `list_tables`, `evaluate_dax`)
- Sets a pandas display option so long text columns are easier to read
- Defines several helper functions, including:
  - `slugify(...)` – turns a name into a safe file name
  - `dax_name(table, column)` and `meas(name)` – build valid DAX references like `'Table'[Column]` and `[Measure]`
  - `safe_call(...)` – safely calls Fabric metadata functions and returns an empty table if there is an error
  - `eval_dax(...)` – runs a DAX query against the semantic model and records the query for later documentation
  - Utility functions for:
    - Renaming columns into a simpler form
    - Detecting which fields are hidden
    - Classifying measures (sales, profit, percentage, etc.)
    - Labeling columns in a user‑friendly way
    - Formatting numeric values into readable text (currency, percentages, etc.)
  - `add_test(...)` – records each generated ground truth test question and its expected answer

Run this cell **once per session** before running later steps.


````python
# ============================================================
# 2. Imports and helper functions
# ============================================================
# %pip install -U semantic-link openpyxl
import json, os, re
from datetime import datetime, timezone
from typing import Any, List, Optional
import pandas as pd
import sempy.fabric as fabric
import tempfile
pd.set_option("display.max_colwidth", 200)

def slugify(v):
    return re.sub(r"[^A-Za-z0-9]+", "_", str(v).strip()).strip("_").lower() or "semantic_model"

def dax_name(table, column):
    return "'" + str(table).replace("'", "''") + "'[" + str(column).replace("]", "]]" ) + "]"

def meas(name):
    return "[" + str(name).replace("]", "]]") + "]"

def safe_call(fn, *args, **kwargs):
    try: return getattr(fabric, fn)(*args, **kwargs)
    except Exception as e:
        print(f"WARN fabric.{fn}: {e}"); return pd.DataFrame()

def eval_dax(dax, qid, purpose):
    validation_queries.append({"query_id": qid, "purpose": purpose, "validation_query": dax})
    try:
        return fabric.evaluate_dax(dataset=SEMANTIC_MODEL_NAME, dax_string=dax, workspace=WORKSPACE_NAME) if WORKSPACE_NAME else fabric.evaluate_dax(dataset=SEMANTIC_MODEL_NAME, dax_string=dax)
    except Exception as e:
        print(f"WARN {qid}: {e}"); return pd.DataFrame()

def norm_cols(df):
    if df.empty: return df
    out=df.copy(); cols=[]
    for c in out.columns:
        s=str(c); s=re.sub(r"^.*\[", "", s).replace("]", "").replace("[", "").strip(); cols.append(s)
    out.columns=cols; return out

def first_col(df, candidates):
    if df.empty: return None
    low={str(c).lower():c for c in df.columns}
    for x in candidates:
        if x.lower() in low: return low[x.lower()]
    for c in df.columns:
        if any(x.lower() in str(c).lower() for x in candidates): return c
    return None

def hidden(row):
    for k in ["Hidden","IsHidden","isHidden","Is Hidden","Object Hidden"]:
        if k in row.index:
            v=row[k]
            return str(v).strip().lower() in ["true","yes","1"] if isinstance(v,str) else bool(v)
    return False

def measure_class(n):
    n=str(n).lower()
    if any(x in n for x in ["sales","revenue","premium","amount"]): return "primary_value"
    if "profit" in n or "margin" in n: return "profitability"
    if "discount" in n: return "discount"
    if "%" in n or "percent" in n or "rate" in n: return "percentage"
    if any(x in n for x in ["count","qty","quantity","volume"]): return "count_or_quantity"
    return "general"

def col_role(table, col, dtype=""):
    text=f"{table} {col} {dtype}".lower()
    if any(x in text for x in ["date","year","month","quarter","fiscal"]): return "date"
    if any(x in text for x in ["country","region","state","province","city","geo","territory"]): return "geography"
    if any(x in text for x in ["category","subcategory","product","item","sku"]): return "product"
    if any(x in text for x in ["segment","customer","client","account","industry"]): return "customer"
    return "dimension"

def label(table, col):
    s=re.sub(r"[_\-]+", " ", f"{table} {col}")
    s=re.sub(r"\b(dim|fact|tbl|vw)\b", "", s, flags=re.I)
    return re.sub(r"\s+", " ", s).strip()

def fmt(v, meta=None, pct=False):
    if pd.isna(v): return "blank"
    if isinstance(v,str): return v
    name="" if meta is None else str(meta.get("_measure_name", meta.get("Name", ""))).lower()
    form="" if meta is None else str(meta.get("FormatString", meta.get("Format String", ""))).lower()
    if pct or "%" in form or any(x in name for x in ["percent","margin","rate"]):
        v=float(v); v=v*100 if abs(v)<=1 else v; return f"{v:,.2f}%"
    if "$" in form or "currency" in form or any(x in name for x in ["sales","revenue","profit","amount","cost","premium"]):
        return f"${float(v):,.0f}"
    v=float(v); return f"{v:,.0f}" if abs(v-round(v))<1e-6 else f"{v:,.2f}"

def add_test(difficulty, category, question, expected, answer_type, filters, qid, notes):
    if len(tests) >= MAX_TESTS: return
    tests.append({"test_id":f"GT{len(tests)+1:03d}","difficulty":difficulty,"evaluation_category":category,"question":question,"expected_answer":expected,"expected_answer_type":answer_type,"important_filters_or_grouping":filters,"validation_dax_or_query_used":qid,"evaluation_notes":notes})
````

## Step 3 – Inspect the semantic model and choose measures/dimensions

In this step, the notebook **reads metadata** from the semantic model and decides which fields are most useful for evaluation.

### What this cell does

1. Calls Fabric / Power BI metadata APIs (via `sempy.fabric`) to get:
   - Tables
   - Columns
   - Measures
   - Relationships
2. Identifies which measures and columns are **visible** (not hidden).
3. Classifies measures into groups such as:
   - Primary value (e.g., sales, revenue)
   - Profitability (e.g., profit, margin)
   - Percentage / rate
   - Counts / quantities
4. Picks a small set of **key measures** to focus on (for example, `Total Sales`, `Profit`, etc.).
5. Scans dimension columns and filters out columns that are less useful for natural questions (IDs, keys, GUIDs, technical columns).
6. Builds a list of **candidate dimensions** (such as product, customer, date, geography) that will be used in questions.

### How to use this step

- Simply **run the cell**.
- Check the output tables:
  - The first output shows the **selected measures** and their classes.
  - The second output shows the **candidate dimension columns**.

If the results do not look right (e.g., the wrong measures are selected), you can:

- Adjust the classification logic in the helper functions (advanced), or
- Change the limits in Step 1 (for example, increase `MAX_MEASURES_FOR_KPIS`).


````python
# ============================================================
# 3. Inspect metadata and select candidates
#
# This cell connects to the semantic model and pulls out basic
# information (tables, columns, measures, relationships). It then:
#   - Finds the right column names in the metadata tables
#   - Filters to only *visible* measures
#   - Classifies measures into simple categories (sales, profit, etc.)
#   - Picks a small set of "key" measures
#   - Builds a list of useful dimension columns (ignoring IDs / keys)
#
# You normally just **run** this cell and review the printed output.
# ============================================================

# Build an optional workspace argument (only used if WORKSPACE_NAME is set)
workspace_arg = {"workspace": WORKSPACE_NAME} if WORKSPACE_NAME else {}

# Call sempy.fabric helpers to read semantic model metadata
# Each call returns a pandas DataFrame with information from the model
tables_df = safe_call("list_tables", SEMANTIC_MODEL_NAME, **workspace_arg)
columns_df = safe_call("list_columns", SEMANTIC_MODEL_NAME, **workspace_arg)
measures_df = safe_call("list_measures", SEMANTIC_MODEL_NAME, **workspace_arg)
relationships_df = safe_call("list_relationships", SEMANTIC_MODEL_NAME, **workspace_arg)
print("tables", tables_df.shape, "columns", columns_df.shape, "measures", measures_df.shape, "relationships", relationships_df.shape)

# Work out which column names to use in the metadata DataFrames
# (different models may use slightly different column labels)
m_col = first_col(measures_df, ["Name", "Measure", "Measure Name"])
t_col = first_col(columns_df, ["Table Name", "Table", "TableName"])
c_col = first_col(columns_df, ["Column Name", "Name", "Column"])
dt_col = first_col(columns_df, ["Data Type", "DataType", "Type"])

# If we cannot find the basic metadata columns, stop early with a clear error
if not m_col or not t_col or not c_col:
    raise ValueError("Could not identify required metadata columns.")

# -----------------------------------------------------------------
# Select visible measures and classify them
# -----------------------------------------------------------------

# Keep only measures that are not hidden in the model
visible_measures = measures_df[~measures_df.apply(hidden, axis=1)].copy()

# Standardize the measure name into a helper column
visible_measures["_measure_name"] = visible_measures[m_col].astype(str)

# Classify each measure into a simple type (primary_value, profitability, etc.)
visible_measures["_class"] = visible_measures["_measure_name"].apply(measure_class)

# Rank the classes so we can pick the "most important" measures first
rank = {
    "primary_value": 1,
    "profitability": 2,
    "percentage": 3,
    "discount": 4,
    "count_or_quantity": 5,
    "general": 9,
}
visible_measures["_rank"] = visible_measures["_class"].map(rank).fillna(9)

# Choose the top N measures (controlled by MAX_MEASURES_FOR_KPIS)
selected_measures = visible_measures.sort_values(["_rank", "_measure_name"]).head(MAX_MEASURES_FOR_KPIS)

# Remember the primary (first) measure and optional profit / percentage measures
primary_measure_row = selected_measures.iloc[0]
primary_measure = primary_measure_row["_measure_name"]

profit_rows = selected_measures[selected_measures["_class"].eq("profitability")]
profit_measure_row = profit_rows.iloc[0] if len(profit_rows) else None
profit_measure = profit_measure_row["_measure_name"] if profit_measure_row is not None else None

pct_rows = selected_measures[selected_measures["_class"].isin(["percentage", "discount", "profitability"])]
pct_measure_row = pct_rows.iloc[0] if len(pct_rows) else None
pct_measure = pct_measure_row["_measure_name"] if pct_measure_row is not None else None

# -----------------------------------------------------------------
# Build a list of candidate dimension columns
# -----------------------------------------------------------------

candidate_dims = []

# Loop through all *visible* columns and keep only those that look like
# useful dimensions for natural-language questions (e.g., Product, Year).
for _, r in columns_df[~columns_df.apply(hidden, axis=1)].iterrows():
    table, col = str(r[t_col]), str(r[c_col])
    dtype = str(r[dt_col]) if dt_col else ""
    lc = col.lower()

    # Exclude technical columns that are unlikely to help with questions
    if lc.endswith("id") or lc in ["id", "key", "index"] or "guid" in lc:
        continue
    if any(x in lc for x in ["row", "hash", "timestamp", "modified", "created", "url"]):
        continue

    # Keep a simple description of each remaining candidate dimension
    candidate_dims.append(
        {
            "table": table,
            "column": col,
            "data_type": dtype,
            "role": col_role(table, col, dtype),  # e.g., date, product, customer
            "label": label(table, col),           # human-friendly label
        }
    )

candidate_dims_df = pd.DataFrame(candidate_dims)

print("Primary measure:", primary_measure)

# Show a small summary of which measures and dimensions were selected
display(selected_measures[["_measure_name", "_class"]])
display(candidate_dims_df.head(50))
````

## Step 4 – Profile the model and automatically generate test questions

This is the **core** step where the notebook analyzes the semantic model and builds a set of ground truth tests.

### What this cell does

1. **Profiles dimensions**
   - For each selected dimension column (like Product, Customer, Year):
     - Runs a DAX query to see distinct values and their primary measure (for example, total sales).
     - Keeps dimensions with a manageable number of distinct values.

2. **Computes overall KPIs**
   - Builds a single row with the main measures (e.g., overall sales, profit, margin).
   - Creates simple questions like: _"What is Total Sales for the full dataset?"_

3. **Builds date‑based tests (if a date column exists)**
   - Finds a Year and possibly a Month column.
   - Creates questions such as:
     - _"What years are available in the dataset?"_
     - _"What was Total Sales in 2019?"_
     - _"Which year had higher Total Sales, 2020 or 2021, and by how much?"_

4. **Creates ranking tests by dimension**
   - For non‑date dimensions (Product, Customer, etc.), generates questions like:
     - _"Which product had the highest Total Sales?"_
     - _"Which product had the lowest Total Sales?"_
     - _"Which customer segment had the highest profit margin?"_

5. **Generates Top‑N and Bottom‑N tests**
   - For selected dimensions, builds sorted lists such as:
     - _"What are the top 3 products by Total Sales?"_
     - _"What are the bottom 3 customers by Profit?"_

6. **Creates multi‑hop tests**
   - Combines two dimensions in a single question, for example:
     - _"Within Region X, which Product had the highest Total Sales?"_
     - _"Within Region X and Product Y, what were Total Sales and Profit?"_

All tests are stored in a Python list called `tests`, and each DAX query used is stored in `validation_queries`.

### How to use this step

- Run the cell.
- At the end, it prints how many tests were generated and shows a preview table.

If **no tests** are generated, check that:

- The semantic model has visible measures and dimensions.
- You have permissions to read the model.
- The configuration in Step 1 is correct.


````python
# ============================================================
# 4. Profile model and generate ground truth tests
#
# This cell:
#   1. Profiles each candidate dimension (e.g., Product, Customer, Year)
#      by running DAX queries against the semantic model.
#   2. Computes overall KPI values for the selected measures
#      (e.g., total sales, profit, margin for the full dataset).
#   3. Builds different types of ground truth questions, such as:
#        - Overall KPI questions (no filters).
#        - Yearly and monthly trend questions.
#        - Ranking questions by dimension (top / bottom item).
#        - Top-N / Bottom-N lists.
#        - Multi-hop questions that combine two dimensions.
#
# All generated tests are stored in the `tests` list.
# Each DAX query used is stored in `validation_queries` so that
# you can always see exactly how the ground truth numbers were computed.
# ============================================================

# Start with empty containers for:
#   - validation_queries: the DAX queries used to compute the answers
#   - tests: the natural-language questions + expected answers
#   - profiled: summary info for dimensions that we successfully profiled
validation_queries = []
tests = []
profiled = []

# ------------------------------------------------------------------
# 1) Profile each candidate dimension
# ------------------------------------------------------------------
# For each candidate dimension column, we:
#   - Build a DAX query that returns distinct values and primary measure
#   - Limit to MAX_DISTINCT_VALUES_FOR_DIMENSION + 1 rows so that
#     we avoid huge dimensions.
#   - Keep only dimensions that return a "small enough" number of
#     distinct values (so they are practical for ranking questions).
for _, d in candidate_dims_df.head(MAX_DIMENSION_COLUMNS * 3).iterrows():
    qid = f"profile_dim_{len(profiled)+1:03d}"  # unique query id
    dn = dax_name(d["table"], d["column"])     # e.g. 'Product'[Category]

    # DAX: profile this dimension by showing distinct values and
    # the primary measure for each value, ordered by measure desc.
    dax = f"""
    EVALUATE
    TOPN(
        {MAX_DISTINCT_VALUES_FOR_DIMENSION+1},
        SUMMARIZECOLUMNS(
            {dn},
            "__PrimaryMeasure", {meas(primary_measure)}
        ),
        [__PrimaryMeasure], DESC,
        {dn}, ASC
    )
    ORDER BY [__PrimaryMeasure] DESC
    """

    # Run the DAX query and normalize the column names
    df = norm_cols(eval_dax(dax, qid, f"Profile {d['label']}"))

    # Keep only dimensions that have a manageable number of distinct values
    if not df.empty and len(df) <= MAX_DISTINCT_VALUES_FOR_DIMENSION:
        x = d.to_dict()
        x["returned_values"] = len(df)
        profiled.append(x)

# Turn the profiled dimensions into a DataFrame for further use
profiled_dims_df = pd.DataFrame(profiled)

# If we have any profiled dimensions, rank them by role and size so that
# we can prioritize the most useful ones (dates, products, customers, etc.).
if not profiled_dims_df.empty:
    role_order = {"date": 1, "product": 2, "customer": 3, "geography": 4, "dimension": 5}
    profiled_dims_df["_role_order"] = profiled_dims_df["role"].map(role_order).fillna(9)
    profiled_dims_df = profiled_dims_df.sort_values(
        ["_role_order", "returned_values", "label"]
    ).head(MAX_DIMENSION_COLUMNS)

print("Profile Dims --------------------------------------")
display(profiled_dims_df.drop(columns=["_role_order"], errors="ignore"))

# ------------------------------------------------------------------
# 2) Compute overall KPI values (no filters)
# ------------------------------------------------------------------
# Build a DAX ROW expression that returns one row containing all
# selected measures for the entire dataset.
row_parts = [
    f'"{r["_measure_name"]}", {meas(r["_measure_name"])}'
    for _, r in selected_measures.iterrows()
]
qid = "query_overall_kpis"
row_parts_str = ",\n    ".join(row_parts)

# Example DAX (conceptually):
# EVALUATE
# ROW(
#   "Total Sales", [Total Sales],
#   "Total Profit", [Total Profit]
# )

dax = f"""
EVALUATE
ROW(
    {row_parts_str}
)
"""

print("Overall KPI values ---------------------------------------")
overall = norm_cols(eval_dax(dax, qid, "Overall KPI values"))
display(overall)

# For each selected measure, create a simple "overall KPI" test.
if not overall.empty:
    for _, m in selected_measures.iterrows():
        mn = m["_measure_name"]
        if mn in overall.columns:
            expected = fmt(overall.loc[0, mn], m)

            # Decide how to label the answer type (number, currency, %)
            if expected.endswith("%"):
                at = "scalar_percentage"
            elif expected.startswith("$"):
                at = "scalar_currency"
            else:
                at = "scalar_number"

            add_test(
                "easy",
                "overall_kpi",
                f"What is {mn} for the full dataset?",
                expected,
                at,
                "No filters; whole model",
                qid,
                "Baseline KPI check using a visible explicit measure.",
            )

# ------------------------------------------------------------------
# 3) Helper to find useful date-like dimensions (Year, Month)
# ------------------------------------------------------------------

def find_dim(tokens):
    """Return the first profiled dimension whose label/text contains
    any of the given tokens (used to find Year/Month fields)."""

    if profiled_dims_df.empty:
        return None

    for _, d in profiled_dims_df.iterrows():
        text = f"{d['table']} {d['column']} {d['label']}".lower()
        if any(t in text for t in tokens):
            return d.to_dict()
    return None

# Try to locate a Year and Month dimension from the profiled dimensions
year_dim = find_dim(["year", "fiscal year"])
month_dim = find_dim(["yearmonth", "year month", "month"])

# ------------------------------------------------------------------
# 4) Year-based tests (if a year dimension exists)
# ------------------------------------------------------------------
if year_dim:
    dn = dax_name(year_dim["table"], year_dim["column"])

    # We'll compute the primary measure and (if available) profit measure by year
    measures = [primary_measure] + (
        [profit_measure] if profit_measure and profit_measure != primary_measure else []
    )

    mp = ",\n    ".join([f'"{m}", {meas(m)}' for m in measures])
    qid = "query_year_trend"

    dax = f"""
    EVALUATE
    SUMMARIZECOLUMNS(
        {dn},
        {mp}
    )
    ORDER BY {dn} ASC
    """

    print("Year Trend ---------------------------------------")
    ydf = norm_cols(eval_dax(dax, qid, "Year trend"))
    display(ydf)

    if not ydf.empty:
        # Pick the year column (fall back to the first column if needed)
        yc = year_dim["column"] if year_dim["column"] in ydf.columns else ydf.columns[0]
        years = list(ydf[yc].dropna())

        # 4a) Question: "What years are available in the dataset?"
        if years:
            add_test(
                "easy",
                "date_coverage",
                "What years are available in the dataset?",
                f"{min(years)} to {max(years)}",
                "scalar_text",
                f"Minimum and maximum of {year_dim['label']}",
                qid,
                "Time coverage check.",
            )

        # 4b) For the first few years, ask for the KPI values per year
        for _, r in ydf.head(5).iterrows():
            pieces = [
                f"{m}: {fmt(r[m], selected_measures[selected_measures['_measure_name'].eq(m)].iloc[0])}"
                for m in measures
                if m in ydf.columns
            ]
            add_test(
                "medium",
                "yearly_trend",
                f"What was {' and '.join(measures)} in {r[yc]}?",
                "; ".join(pieces),
                "record",
                f"Filter {year_dim['label']} = {r[yc]}",
                qid,
                "Single-period filtered KPI retrieval.",
            )

        # 4c) Compare two years on the primary measure and ask which is higher
        if len(ydf) >= 2 and primary_measure in ydf.columns:
            s = ydf.sort_values(primary_measure, ascending=False)
            diff = float(s.iloc[0][primary_measure]) - float(s.iloc[1][primary_measure])
            add_test(
                "medium",
                "yearly_comparison",
                f"Which {year_dim['label']} had higher {primary_measure}, {s.iloc[0][yc]} or {s.iloc[1][yc]}, and by how much?",
                f"{s.iloc[0][yc]} was higher by {fmt(diff, primary_measure_row)}",
                "scalar_text",
                f"Compare {primary_measure} across {year_dim['label']}",
                qid,
                "Comparison and subtraction across grouped results.",
            )

# ------------------------------------------------------------------
# 5) Month-based tests (if a month dimension exists)
# ------------------------------------------------------------------
if month_dim:
    dn = dax_name(month_dim["table"], month_dim["column"])

    measures = [primary_measure] + (
        [profit_measure] if profit_measure and profit_measure != primary_measure else []
    )

    mp = ",\n    ".join([f'"{m}", {meas(m)}' for m in measures])
    qid = "query_month_trend"

    dax = f"""
EVALUATE
SUMMARIZECOLUMNS(
    {dn},
    {mp}
)
ORDER BY {dn} ASC
"""

    mdf = norm_cols(eval_dax(dax, qid, "Month trend"))

    if not mdf.empty and primary_measure in mdf.columns:
        # Pick the month column (fall back to first column if needed)
        mc = month_dim["column"] if month_dim["column"] in mdf.columns else mdf.columns[0]

        # Find the months with highest and lowest primary measure
        high = mdf.sort_values(primary_measure, ascending=False).iloc[0]
        low = mdf.sort_values(primary_measure).iloc[0]

        # 5a) Month with highest primary measure
        add_test(
            "medium",
            "monthly_trend",
            f"Which {month_dim['label']} had the highest {primary_measure}?",
            f"{high[mc]} with {fmt(high[primary_measure], primary_measure_row)}",
            "record",
            f"Group by {month_dim['label']}",
            qid,
            "Global maximum over monthly series.",
        )

        # 5b) Month with lowest primary measure
        add_test(
            "medium",
            "monthly_trend",
            f"Which {month_dim['label']} had the lowest {primary_measure}?",
            f"{low[mc]} with {fmt(low[primary_measure], primary_measure_row)}",
            "record",
            f"Group by {month_dim['label']}",
            qid,
            "Global minimum over monthly series.",
        )

# ------------------------------------------------------------------
# 6) Dimension ranking tests (non-date dimensions)
# ------------------------------------------------------------------

# Keep only non-date dimensions for general ranking questions
non_date = (
    profiled_dims_df[profiled_dims_df["role"].ne("date")].to_dict("records")
    if not profiled_dims_df.empty
    else []
)

for i, d in enumerate(non_date):
    dn = dax_name(d["table"], d["column"])

    # Start with primary measure, optionally add profit and percentage
    measures = [primary_measure]
    if profit_measure and profit_measure != primary_measure:
        measures.append(profit_measure)
    if pct_measure and pct_measure not in measures:
        measures.append(pct_measure)

    mp = ",\n    ".join([f'"{m}", {meas(m)}' for m in measures])
    qid = f"query_rank_{i+1:03d}"

    dax = f"""
EVALUATE
SUMMARIZECOLUMNS(
    {dn},
    {mp}
)
ORDER BY [{primary_measure}] DESC
"""

    df = norm_cols(eval_dax(dax, qid, f"Ranking by {d['label']}"))
    if df.empty or primary_measure not in df.columns:
        continue

    # Dimension label column (fallback to first column if needed)
    dc = d["column"] if d["column"] in df.columns else df.columns[0]

    # Highest and lowest values for primary measure
    top = df.sort_values(primary_measure, ascending=False).iloc[0]
    low = df.sort_values(primary_measure).iloc[0]

    # 6a) Highest primary measure for this dimension
    add_test(
        "easy" if d["returned_values"] <= 10 else "medium",
        f"{d['role']}_ranking",
        f"Which {d['label']} had the highest {primary_measure}?",
        f"{top[dc]} with {fmt(top[primary_measure], primary_measure_row)}",
        "record",
        f"Group by {d['label']}",
        qid,
        "Dimension ranking benchmark.",
    )

    # 6b) Lowest primary measure for this dimension
    add_test(
        "medium",
        f"{d['role']}_ranking",
        f"Which {d['label']} had the lowest {primary_measure}?",
        f"{low[dc]} with {fmt(low[primary_measure], primary_measure_row)}",
        "record",
        f"Group by {d['label']}",
        qid,
        "Bottom ranking benchmark.",
    )

    # 6c) Percentage-based ranking if a percentage measure exists
    if pct_measure and pct_measure in df.columns:
        pct_meta = selected_measures[selected_measures["_measure_name"].eq(pct_measure)].iloc[0]
        ptop = df.sort_values(pct_measure, ascending=False).iloc[0]
        add_test(
            "medium",
            f"{d['role']}_percentage",
            f"Which {d['label']} had the highest {pct_measure}?",
            f"{ptop[dc]} at {fmt(ptop[pct_measure], pct_meta, True)}",
            "record",
            f"Group by {d['label']}",
            qid,
            "Percentage or margin ranking benchmark.",
        )

# ------------------------------------------------------------------
# 7) Top-N and Bottom-N tests for selected rankable dimensions
# ------------------------------------------------------------------

# Restrict to dimensions that look like product or customer-like entities
rankable = [
    d
    for d in non_date
    if any(t in f"{d['role']} {d['label']}".lower() for t in ["product", "item", "sku", "customer", "client", "account"])
]

for d in rankable[:3]:  # only use first few to avoid generating too many tests
    dn = dax_name(d["table"], d["column"])
    # Count how many top-N queries we already have to form a unique id
    qid = f"query_topn_{len([q for q in validation_queries if str(q['query_id']).startswith('query_topn_')])+1:03d}"

    dax = f"""
EVALUATE
TOPN(
    {max(TOP_N,10)},
    SUMMARIZECOLUMNS(
        {dn},
        "{primary_measure}", {meas(primary_measure)}
    ),
    [{primary_measure}], DESC,
    {dn}, ASC
)
ORDER BY [{primary_measure}] DESC, {dn} ASC
"""

    df = norm_cols(eval_dax(dax, qid, f"Top {d['label']} by {primary_measure}"))

    if not df.empty and primary_measure in df.columns:
        dc = d["column"] if d["column"] in df.columns else df.columns[0]

        # Build a readable Top-N list: "1. Product A – $1,234; 2. Product B – $987; ..."
        items = [
            f"{n}. {r[dc]} – {fmt(r[primary_measure], primary_measure_row)}"
            for n, (_, r) in enumerate(df.head(TOP_N).iterrows(), 1)
        ]

        add_test(
            "hard",
            "top_n",
            f"What are the top {TOP_N} {d['label']} values by {primary_measure}?",
            "; ".join(items),
            "ranked_list",
            f"Rank {d['label']} by {primary_measure} descending",
            qid,
            "Top-N ranking benchmark with order sensitivity.",
        )

    # Optional: Bottom-N by profit, if a profit measure is available
    if profit_measure:
        qid2 = f"query_bottomn_{len([q for q in validation_queries if str(q['query_id']).startswith('query_bottomn_')])+1:03d}"

        dax2 = f"""
    EVALUATE
TOPN(
    {max(TOP_N,10)},
    SUMMARIZECOLUMNS(
        {dn},
        "{profit_measure}", {meas(profit_measure)}
    ),
    [{profit_measure}], ASC,
    {dn}, ASC
)
ORDER BY [{profit_measure}] ASC, {dn} ASC
"""

        df2 = norm_cols(eval_dax(dax2, qid2, f"Bottom {d['label']} by {profit_measure}"))

        if not df2.empty and profit_measure in df2.columns:
            dc = d["column"] if d["column"] in df2.columns else df2.columns[0]

            items = [
                f"{n}. {r[dc]} – {fmt(r[profit_measure], profit_measure_row)}"
                for n, (_, r) in enumerate(df2.head(TOP_N).iterrows(), 1)
            ]

            add_test(
                "hard",
                "bottom_n",
                f"What are the bottom {TOP_N} {d['label']} values by {profit_measure}?",
                "; ".join(items),
                "ranked_list",
                f"Rank {d['label']} by {profit_measure} ascending",
                qid2,
                "Bottom-N ranking benchmark with order sensitivity.",
            )

# ------------------------------------------------------------------
# 8) Multi-hop tests (two dimensions together)
# ------------------------------------------------------------------

# If we have at least two non-date dimensions, create questions that:
#   - Filter by one dimension (e.g., Region)
#   - Rank or filter by another dimension (e.g., Product inside Region)
if len(non_date) >= 2:
    d1, d2 = non_date[0], non_date[1]
    dn1 = dax_name(d1["table"], d1["column"])
    dn2 = dax_name(d2["table"], d2["column"])

    measures = [primary_measure] + (
        [profit_measure] if profit_measure and profit_measure != primary_measure else []
    )

    mp = ",\n    ".join([f'"{m}", {meas(m)}' for m in measures])
    qid = "query_dimension_by_dimension"

    dax = f"""
EVALUATE
SUMMARIZECOLUMNS(
    {dn1},
    {dn2},
    {mp}
)
ORDER BY {dn1} ASC, [{primary_measure}] DESC
"""

    df = norm_cols(eval_dax(dax, qid, f"{d1['label']} by {d2['label']}"))

    if not df.empty and primary_measure in df.columns:
        # Identify the columns for the two dimensions in the result
        c1 = d1["column"] if d1["column"] in df.columns else df.columns[0]
        c2 = d2["column"] if d2["column"] in df.columns else df.columns[1]

        # Choose a specific value of the first dimension (e.g., a region)
        gv = df[c1].dropna().iloc[0]

        # Filter to that value and find the row with the highest primary measure
        scoped = df[df[c1].eq(gv)].sort_values(primary_measure, ascending=False)
        top = scoped.iloc[0]

        # 8a) Question: within this first dimension value, which second
        # dimension value has the highest primary measure?
        add_test(
            "hard",
            "multi_hop_filters",
            f"Within {d1['label']} {gv}, which {d2['label']} had the highest {primary_measure}?",
            f"{top[c2]} with {fmt(top[primary_measure], primary_measure_row)}",
            "record",
            f"Filter {d1['label']} = {gv}; group by {d2['label']}",
            qid,
            "Multi-hop test across two model dimensions.",
        )

        # 8b) Question: within this combination (dim1 value + dim2 value),
        # what are the values of one or more measures?
        pieces = [
            f"{m}: {fmt(top[m], selected_measures[selected_measures['_measure_name'].eq(m)].iloc[0])}"
            for m in measures
            if m in df.columns
        ]

        add_test(
            "hard",
            "multi_hop_filters",
            f"Within {d1['label']} {gv}, what were {' and '.join(measures)} for {d2['label']} {top[c2]}?",
            "; ".join(pieces),
            "record",
            f"Filter {d1['label']} = {gv}, {d2['label']} = {top[c2]}",
            qid,
            "Multi-hop filtered aggregation.",
        )

print("Generated Tests ---------------------------")
print("Generated tests:", len(tests))
display(pd.DataFrame(tests))
````

## Step 5 – Save the ground truth dataset to your Lakehouse

In this final step, the notebook takes all of the generated tests and writes them out as files so you can reuse them in other tools and pipelines.

### What this cell does

1. Converts the in‑memory `tests` list into a pandas DataFrame (`tests_df`).
2. Stops with a clear error if **no tests** were generated in Step 4.
3. Cleans up the tests:
   - Removes duplicate questions.
   - Enforces the `MAX_TESTS` limit.
   - Renumbers `test_id` in order (GT001, GT002, ...).
4. Builds helper tables:
   - `validation_queries_df` – all DAX queries used to compute ground truth.
   - `question_candidates_df` – potential questions based on measures and dimensions.
   - `coverage_df` – how many tests were created per category and difficulty.
   - `metadata` – context about the semantic model, generation time, and selected measures/dimensions.
5. Writes the main outputs **to a temporary local folder**, then copies them into the Lakehouse:
   - CSV: `<OUTPUT_BASENAME>.csv`
   - JSON: `<OUTPUT_BASENAME>.json`

   These files are stored under the folder:

   ```text
   Files/ground_truth
   ```

6. Prints a summary of what was written and shows coverage and tests in the notebook output.

### After running this step

- Open the **Lakehouse** attached to this notebook (for example `FabricDataAgentBenchmarks`).
- Browse to `Files/ground_truth` and download or reference:
  - The CSV file for use in Excel, Power BI, or other tools.
  - The JSON file for automated evaluation or integration.

You can now use this dataset to **evaluate and benchmark Fabric data agents** against your semantic model.


````python
# ============================================================
# 5. Finalize and write CSV, JSON, XLSX
# ============================================================

# Rebuild tests_df from tests
# Stop with a clear ValueError if no tests
tests_df = pd.DataFrame(tests)
if tests_df.empty:
    raise ValueError("No tests were generated. Review metadata, visible measures, and permissions.")

basename = OUTPUT_BASENAME or f"{slugify(SEMANTIC_MODEL_NAME)}_ground_truth"

# Writes CSV + JSON to a temp local folder
local_dir = tempfile.mkdtemp(prefix="ground_truth_")
local_csv  = os.path.join(local_dir, f"{basename}.csv")
local_json = os.path.join(local_dir, f"{basename}.json")
# local_xlsx = os.path.join(local_dir, f"{basename}.xlsx")


# Cleanse tests
tests_df = tests_df.drop_duplicates(subset=["question"], keep="first").head(MAX_TESTS).reset_index(drop=True)
tests_df["test_id"]=[f"GT{i+1:03d}" for i in range(len(tests_df))]
validation_queries_df=pd.DataFrame(validation_queries).drop_duplicates(subset=["query_id"], keep="first")
question_candidates=[]
for _,m in selected_measures.iterrows(): question_candidates.append({"candidate_type":"measure","name":m['_measure_name'],"classification":m['_class'],"suggested_question":f"What is {m['_measure_name']} for the full dataset?"})
for _,d in profiled_dims_df.iterrows() if not profiled_dims_df.empty else []: question_candidates.append({"candidate_type":"dimension","name":f"{d['table']}[{d['column']}]","classification":d['role'],"suggested_question":f"Which {d['label']} had the highest {primary_measure}?"})
question_candidates_df=pd.DataFrame(question_candidates)
coverage_df=tests_df.groupby(["evaluation_category","difficulty"]).size().reset_index(name="test_count")
metadata={"dataset_name":f"{SEMANTIC_MODEL_NAME} Fabric Data Agent Ground Truth Dataset","source_semantic_model":SEMANTIC_MODEL_NAME,"workspace":WORKSPACE_NAME or "Current workspace context","generated_for":"Fabric data agent evaluation","generated_utc":datetime.now(timezone.utc).isoformat(),"row_count":int(len(tests_df)),"primary_measure":primary_measure,"selected_measures":selected_measures['_measure_name'].tolist(),"selected_dimensions":profiled_dims_df.drop(columns=['_role_order'], errors='ignore').to_dict('records') if not profiled_dims_df.empty else [],"known_omissions":["Only metadata and DAX queries accessible to the executing user are included.","Hidden fields are skipped when visibility metadata is available.","Generated questions are heuristic and should be reviewed by a model owner or SME before formal scoring."]}

basename=OUTPUT_BASENAME or f"{slugify(SEMANTIC_MODEL_NAME)}_ground_truth"

# Write CSV/JSON locally first
tests_df.to_csv(local_csv,index=False)
with open(local_json,"w",encoding="utf-8") as f: json.dump({"metadata":metadata,"tests":tests_df.to_dict('records'),"validation_queries":validation_queries_df.to_dict('records'),"question_candidates":question_candidates_df.to_dict('records')},f,indent=2,default=str)
metadata_df=pd.DataFrame([{"key":k,"value":json.dumps(v,default=str) if isinstance(v,(list,dict)) else v} for k,v in metadata.items()])

# Ensure Lakehouse folder exists
OUTPUT_FOLDER = "Files/ground_truth"
notebookutils.fs.mkdirs(OUTPUT_FOLDER)

# Capture directory state BEFORE copy
# before_listing = notebookutils.fs.ls(OUTPUT_FOLDER)
# before_paths = {entry["path"] for entry in before_listing}
before_listing = notebookutils.fs.ls(OUTPUT_FOLDER)
before_paths = {entry.path for entry in before_listing}

# Copy into Lakehouse Files
notebookutils.fs.fastcp(f"file:{local_csv}",  f"{OUTPUT_FOLDER}/{basename}.csv")
notebookutils.fs.fastcp(f"file:{local_json}", f"{OUTPUT_FOLDER}/{basename}.json")

# Capture directory state AFTER copy
# after_listing = notebookutils.fs.ls(OUTPUT_FOLDER)
# after_paths = {entry["path"] for entry in after_listing}
after_listing = notebookutils.fs.ls(OUTPUT_FOLDER)
after_paths = {entry.path for entry in after_listing}

added_paths = sorted(after_paths - before_paths)

print("----------------------------------")
print(f"Directory: {OUTPUT_FOLDER}")
print(f"Files before job: {len(before_paths)}")
print(f"Files after job:  {len(after_paths)}")
print(f"Files added:      {len(added_paths)}")
if added_paths:
    print("Added files:")
    for p in added_paths:
        print("   ", p)
else:
    print("No new files added (files may have been overwritten).")

print("----------------------------------")
print("Coverage:")
display(coverage_df)
print("----------------------------------")
print("Tests:")
display(tests_df)
````
